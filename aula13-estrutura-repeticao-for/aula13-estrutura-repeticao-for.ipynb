```python
# Perceba um detalhe,
# eu quero escrever "oi" 5 vezes.

# Em um primeiro momento parece correto,
# mas na saída produz no total 4 "oi"s.
for i in range(1, 5):
    print('oi')

# Isso acontece porque os parâmetros do range()
# não são "começo" e "final" (do 1 ao 5), mas sim
# "começo" e "total" (começando do 1 de 5 espaços),
# porém o python começa contabilizando do 0, e não do 1
```

```python
# Diferença do comando ser com e sem identação
# Não repete o "FIM", ele aparece somente no final.
for i in range(0, 5):
    print('oi')
# Sem identar == estar fora do laço
print('FIM!')

print()

# Repete o "FIM" a cada repetição.
for i in range(0, 5):
    print('oi')
    # Identado == estar dentro do laço
    print('FIM!')
```

```python
# Usando só um parâmetro no range()
# E usando a própria variável de controle do for dentro dele,
# a variável i vai mudando de valor a cada repetição,
# por isso ela funcina como uma contagem.
# Usando um parâmetro só define o total de repetições do início (0).
for i in range(5):
    print(i)

# E preste atenção, como dito anteriormente, a contagem começa no 0.
```

```python
# Contagem regressiva
# O terceiro parâmetro é o andamento da contagem
for i in range(6, 0, -1):
    print(i)

# O total é 6, começa da posição 0, mas para no 1
```

```python
num = int(input('Digite um valor: '))

for i in range(0, num):
    print(i + 1)

# Laço que vai até onde o usuário quiser.
```

```python
inicio = int(input('Início: '))
fim    = int(input('Fim: '   ))
passo  = int(input('Passo: ' ))

for i in range(inicio, fim + 1, passo):
    print(i)

# Perceba que eu inseri fim + 1, porque o user
# pensa que o parâmetro fim é literalmente o valor final,
# é intuitivo, porém está errado, o segundo parãmetro é o
# total de espaços no range().

# Ex.: de "4 a 12" quer dizer em 12 espaços (0 ao 11, total 12)
# começa da posição 4 (0, 1, 2, 3, [4], que é literalmente o valor 4).
 
```

```python
# Ler várias vezes.
for i in range(0, 3):
    num = int(input('Digite um valor: '))
    print(num)
print('fim')
```

```python
# Acumular valores (somador)
soma = 0
for i in range(0, 3):
    num = int(input('Informe um valor: '))
    soma = soma + num
    # Podemos escrever como soma += num,
    # "soma recebe ele mesmo mais num".

print(soma)
```
