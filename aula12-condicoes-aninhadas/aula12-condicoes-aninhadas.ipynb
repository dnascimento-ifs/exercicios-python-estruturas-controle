# Aula 12 - Condições Aninhadas
Uma condição aninhada é quando usamos mais de uma condição na mesma estrutura condicional, ou seja, uma condição dentro da outra condição, esse é o chamado ato de aninhar (ou encadear) condições, e em python temos algumas peculiaridade com sua sintaxe.

- if sempre tem que existir dentro de uma estrutura condicional;
- elif e else são opcionais dentro de uma estrutura condicional;
- dentro de uma mesma estrutura condicional podem existir quantos elifs forem necessários.


```python
# Exemplo 1
nome = str(input('Informe o seu nome: '))

# Condição simples
if nome == 'Gustavo':
    print('Que nome bonito você tem!')

print(f'Tenha um bom dia, {nome}!')
```

```python
# Exemplo 2
nome = str(input('Informe o seu nome: '))

# Condição composta
if nome == 'Gustavo':
    print('Que nome bonito você tem!')
else:
    print('Seu nome é bem normal!')

print(f'Tenha um bom dia, {nome}!')
```

```python
# Exemplo 3
nome = str(input('Informe o seu nome: '))

# Condição aninhada com um elif só (3 caminhos diferentes)
if nome == 'Gustavo':
    print('Que nome bonito você tem!')
elif nome == 'Pedro' or nome == 'Maria' or nome == 'Paulo':
    print('Seu nome é bem comum no Brasil.')
else:
    print('Seu nome é bem normal.')

print(f'Tenha um bom dia, {nome}!')
```

```python
# Exemplo 4
nome = str(input('Informe o seu nome: '))

# Condição aninhada com mais de um elif (4 caminhos diferentes)
# Nela usamos if, elif (else if que é abreviado no python) e else.
if nome == 'Gustavo':
    print('Que nome bonito você tem!')
elif nome == 'Pedro' or nome == 'Maria' or nome == 'Paulo':
    print('Seu nome é bem comum no Brasil.')
    
    # Um exemplo interessante de uso do operador in
elif nome in 'Ana Cláudia Jéssica Juliana':
    print('Belo nome feminino!')

    # Coloque as linhas abaixo da estrutura condicional
    # como comentários e observe que a estrutura funciona
    # sem usar o else.
else:
    print('Seu nome é bem normal.')

print(f'Tenha um bom dia, {nome}!')
```
